```python
from bs4 import BeautifulSoup as bs
import requests as req
class BeautifyPage:

    def check_url(self, url):
        headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'}
        return req.get(url, headers=headers)

    def get_markdown(self, resp, type = 'html'):
        return bs(resp.text, type)

    def get_selector(self, soup, tag, by_class=False, value=None):
        if by_class:
            return soup.find(tag, class_=value)
    
        return soup.find(tag, { "id": value })
```
